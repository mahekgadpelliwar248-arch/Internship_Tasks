Q1
1. Start with a sentence.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Count each word and store its frequency in the dictionary.
5. Return the dictionary.

```python
def word_frequency(sentence):
    result = {}

    for word in sentence.split():
        result[word] = result.get(word, 0) + 1

    return result

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("hello hello world"))
```

Q2
1. Start with a word.
2. Create an empty dictionary.
3. Visit each character in the word.
4. Count each character using the dictionary.
5. Return the dictionary.

```python
def character_frequency(word):
    result = {}

    for char in word:
        result[char] = result.get(char, 0) + 1

    return result

print(character_frequency("banana"))
print(character_frequency("hello"))
```

Q3
1. Start with a word.
2. Count the frequency of every character.
3. Visit the characters in their original order.
4. Return the first character whose frequency is one.
5. If no such character exists, return None.

```python
def first_non_repeating(word):
    frequency = {}

    for char in word:
        frequency[char] = frequency.get(char, 0) + 1

    for char in word:
        if frequency[char] == 1:
            return char

    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))
```

Q4
1. Take two strings.
2. Count the frequency of each character in both strings.
3. Compare the two dictionaries.
4. Return True if they are equal; otherwise, return False.

```python
def are_anagrams(s1, s2):
    return character_frequency(s1) == character_frequency(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
```

Q5
1. Start with a string.
2. Create an empty set.
3. Visit each character from left to right.
4. If the character is already in the set, return it.
5. Otherwise, add it to the set.
6. If no character repeats, return None.

```python
def first_repeating_char(text):
    seen = set()

    for char in text:
        if char in seen:
            return char
        seen.add(char)

    return None

print(first_repeating_char("programming"))
print(first_repeating_char("abc"))
```

Q6
1. Start with a list of numbers.
2. Store all non-zero numbers in a new list.
3. Count the zeros in the original list.
4. Add all zeros to the end.
5. Return the new list.

```python
def move_zeros(nums):
    result = []

    for num in nums:
        if num != 0:
            result.append(num)

    result.extend([0] * nums.count(0))
    return result

print(move_zeros([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros([0, 1, 0, 2]))
```

Q7
1. Start with a list of numbers.
2. Remove duplicate values.
3. Sort the unique numbers in descending order.
4. If at least two numbers exist, return the second number.
5. Otherwise, return None.

```python
def second_highest(nums):
    unique_nums = sorted(set(nums), reverse=True)

    if len(unique_nums) < 2:
        return None

    return unique_nums[1]

print(second_highest([10, 20, 40, 30, 40, 20]))
print(second_highest([5, 5, 5]))
```

Q8
1. Start with a list.
2. Create an empty result list.
3. Visit each element.
4. Add the element only if it is not already in the result.
5. Return the result list.

```python
def remove_duplicates(nums):
    result = []

    for num in nums:
        if num not in result:
            result.append(num)

    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates([1, 1, 2, 3, 2]))
```

Q9
1. Start with a list and the value N.
2. Calculate the sum of numbers from 1 to N.
3. Calculate the sum of the given list.
4. Subtract the list sum from the expected sum.
5. Return the missing number.

```python
def missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(missing_number([1, 2, 4, 5, 6], 6))
print(missing_number([1, 3, 4], 4))
```

Q10
1. Start with a list and a target.
2. Visit each number from left to right.
3. Check earlier numbers to find a pair that sums to the target.
4. Return the first valid pair found.
5. If no pair exists, return an empty list.


```python
def two_sum(nums, target):
    seen = []

    for num in nums:
        needed = target - num

        if needed in seen:
            return [needed, num]

        seen.append(num)

    return []

print(two_sum([2, 7, 11, 15], 9))
print(two_sum([1, 2, 3], 10))
```

Q11
1. Start with a list of numbers.
2. Compare numbers by their absolute values.
3. If two numbers are equally close to zero, prefer the positive number.
4. Return the closest number.

```python
def closest_to_zero(nums):
    return min(nums, key=lambda x: (abs(x), -x))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, 3, 7]))
```

Q12
1. Start with two lists.
2. Create an empty result list.
3. Visit elements of the first list in order.
4. Add an element if it exists in the second list and is not already in the result.
5. Return the result.

```python
def common_elements(list1, list2):
    result = []

    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([2, 2, 3], [2, 3]))
```

Q13
1. Start with a list and K.
2. If the list is empty, return an empty list.
3. Reduce K using the list length.
4. Move the last K elements to the beginning.
5. Return the rotated list.

```python
def rotate_right(nums, k):
    if not nums:
        return []

    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k else nums[:]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 1))
```

Q14
1. Start with a list containing at least two numbers.
2. Add each number to the next number.
3. Track the largest sum.
4. Return the maximum sum.

```python
def max_consecutive_sum(nums):
    largest = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current = nums[i] + nums[i + 1]

        if current > largest:
            largest = current

    return largest

print(max_consecutive_sum([4, 7, 2, 9, 5]))
print(max_consecutive_sum([1, 2, 10, 3]))
```

Q15
1. Start with a list and a target.
2. Check every pair of numbers at different positions.
3. If their sum equals the target, sort the pair.
4. Add the pair only if it is not already in the result.
5. Sort all pairs by their first element.
6. Return the result.

```python
def all_pairs(nums, target):
    result = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])

                if pair not in result:
                    result.append(pair)

    result.sort(key=lambda pair: pair[0])
    return result

print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(all_pairs([1, 2, 3], 20))
```
